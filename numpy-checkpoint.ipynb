```python
import numpy as np

arr = np.arange(1, 13).reshape(3, 4)

# (a) Element in row 2, column 3
print(arr[1, 2])

# (b) Last row using negative indexing
print(arr[-1])

# (c) Last column
print(arr[:, -1])
```
