```python
#13
import numpy as np

x = np.arange(1, 13)

arr = x.reshape(3, 4)
a = x.reshape((12, 1))
b = x[:, np.newaxis]

print(arr)
print(a)
print(b)
print(np.array_equal(a, b))
```

```python
#14
import numpy as np

np.random.seed(0)

a = np.random.randint(0, 10, (2, 3))
b = np.random.randint(0, 10, (2, 3))

v1 = np.vstack((a, b))
v2 = np.concatenate((a, b), axis=0)

h1 = np.hstack((a, b))
h2 = np.concatenate((a, b), axis=1)

print(v1)
print(v2)
print(np.array_equal(v1, v2))

print(h1)
print(h2)
print(np.array_equal(h1, h2))
```
